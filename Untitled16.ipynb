```python
# Define a sample sentence representing factual information for RAG
sentence = "Retrieval-Augmented Generation improves LLM responses using external knowledge."

# Basic word-level tokenization
tokens = sentence.lower().replace(".", "").split()

# Create a simple vocabulary mapping tokens to unique IDs
vocab = {token: idx for idx, token in enumerate(sorted(list(set(tokens))))}

# Convert the sentence tokens into token IDs
token_ids = [vocab[token] for token in tokens]

print(f"Original Sentence: \"{sentence}\"\n")
print(f"Tokens: {tokens}\n")
print(f"Vocabulary Mapping: {vocab}\n")
print(f"Token IDs (Vector Input): {token_ids}")
```
